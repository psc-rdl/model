```python
import gurobipy as gp
from gurobipy import GRB
import numpy as np
import time
import csv

# Initialize model
myProgram = gp.Model("mip1")

## Variables x, y, z binary

# Initialize variables and add them to model
x = myProgram.addVar(vtype=GRB.BINARY, name="x")
y = myProgram.addVar(vtype=GRB.BINARY, name="y")
z = myProgram.addVar(vtype=GRB.BINARY, name="z")
```

```python
# Constraint: x + 2y + 3z <= 4
myProgram.addConstr(x + 2 * y + 3 * z<=4, "c0")

# Add constraint: x + y >= 1
myProgram.addConstr(x + y >= 1, "c1")
```

```python
# Set objective function
myProgram.setObjective(x + y + 2 * z, GRB.MAXIMIZE)

"""
You can also add to the objective function iteratively:
    obj = gp.LinExpr()
    obj += x
    obj += y
    obj += 2 * z
    m.setObjective(obj, GRB.MAXIMIZE)
"""
```

```python
# Optimize model
myProgram.optimize()
```

```python
# Print solutions
for v in myProgram.getVars():
    print('%s %g' % (v.VarName, v.X))

print('Obj: %g' % myProgram.ObjVal)
```
